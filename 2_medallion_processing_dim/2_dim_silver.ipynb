```python
import pyspark.sql.functions as F
```

```python
catalog = "ecommerce"
```

### Brands

```python
df_bronze = spark.table(f"{catalog}.bronze.brz_brands")
df_bronze.show(10)
```

```python
df_bronze = df_bronze.withColumn("brand_name", F.trim(F.col("brand_name")))
df_bronze.show(10)
```

```python
df_bronze = df_bronze.withColumn("brand_code", F.regexp_replace(F.col("brand_code"), "[^A-Za-z0-9]", ""))
df_bronze.show(10)
```

```python
df_bronze.select(F.col("category_code")).distinct().show()
```

```python
df_bronze = df_bronze.replace({"GROCERY": "GRCY", "TOYS": "TOY"}, subset=["category_code"])
df_bronze.show(10)
```

```python
df_bronze.write.format("delta").mode('overwrite').option("mergeschema",True).saveAsTable(f"{catalog}.silver.slv_brands")
```

### Category

```python
df_bronze = spark.table(f"{catalog}.bronze.brz_category")
df_bronze.show(10)
```

```python
df_bronze.groupby("category_code").count().filter(F.col("count")>1).show()
```

```python
df_bronze = df_bronze.dropDuplicates(['category_code'])
df_bronze.show(10)
```

```python
df_bronze = df_bronze.withColumn("category_code", F.upper(F.col("category_code")))
df_bronze.show(10)
```

```python
df_bronze.write.format("delta").mode("overwrite").option("mergeschema",True).saveAsTable(f"{catalog}.silver.slv_category")
```

### Products

```python
df_bronze = spark.table(f"{catalog}.bronze.brz_products")
df_bronze.show(10)
```

```python
df_bronze = df_bronze.withColumn("weight_grams", F.regexp_replace(F.col("weight_grams"), "g", "").cast("int"))
df_bronze.show(10)
```

```python
df_bronze = df_bronze.withColumn(
    "rating_count",
    F.when((F.col("rating_count") < 0) | F.col("rating_count").isNull(), 0).otherwise(F.col("rating_count"))
)
df_bronze.show(10)
```

```python
df_bronze = df_bronze.withColumn("length_cm", F.regexp_replace(F.col("length_cm"), ",", "."))
df_bronze.show(10)
```

```python
df_bronze = df_bronze.withColumn("category_code", F.upper(F.col("category_code"))) \
                     .withColumn("brand_code", F.upper(F.col("brand_code")))
df_bronze.show(10)
```

```python
df_bronze.select(F.col("material")).distinct().show()
```

```python
df_bronze.select(F.col("material")).distinct().show()

material_mapping = {
    "Coton": "Cotton",
    "Ruber": "Rubber",
    "Alumium": "Aluminum",
}

df_bronze = df_bronze.replace(material_mapping, subset=["material"])
df_bronze.show(10)
```

```python
df_bronze.write.\
format("delta").\
option("mergeschema",True).\
mode("overwrite").\
saveAsTable(f"{catalog}.silver.slv_products")
```

### Customers

```python
df_bronze = spark.table(f"{catalog}.bronze.brz_customers")
df_bronze.show(5)
```

```python
df_bronze = df_bronze.withColumn(
    "phone",
    F.regexp_replace(F.col("phone"), "\\..*$", "")
)

df_bronze.show(10)
```

#Handling Null Values

```python
df_bronze.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c)
    for c in df_bronze.columns
]).show()
```

```python
df_bronze = df_bronze.dropna(subset=["customer_id"])
df_bronze.show(10)
```

```python
df_bronze = df_bronze.fillna("Not Available",subset=['Phone'])
df_bronze.show(10)
```

```python
df_bronze.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c)
    for c in df_bronze.columns
]).show()
```

```python
df_bronze.write.\
format("delta").\
option("mergeschema",True).\
mode("overwrite").\
saveAsTable(f"{catalog}.silver.slv_customers")
```

### Calender/Date

```python
df_bronze = spark.table(f"{catalog}.bronze.brz_calendar")
df_bronze.show()
```

```python
df_bronze.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c)
    for c in df_bronze.columns
]).show()
```

```python
df_bronze.groupby("date").count().show()
```

```python
df_bronze.groupby("date").count().filter("count>1").show()
```

```python
df_bronze = df_bronze.dropDuplicates(["date"])
df_bronze.show()
```

```python
df_bronze = df_bronze.withColumn("day_name", F.initcap(F.trim(F.col("day_name"))))
df_bronze.show()
```

```python
df_bronze = df_bronze.withColumn("week_of_year", F.abs(F.col("week_of_year")))
df_bronze.show()
```

```python
df_bronze = (
    df_bronze
    .withColumn("quarter", F.concat(F.lit("Q"), F.col("quarter"), F.lit("-"), F.col("year")))
    .withColumn("week_of_year", F.concat(F.lit("Week"), F.col("week_of_year"), F.lit("-"), F.col("year")))
)

df_bronze.show()
```

```python
# Rename a column
df_bronze = df_bronze.withColumnRenamed("week_of_year", "week")
df_bronze.show()
```

```python
df_bronze.write.\
format("delta").\
option("mergeschema",True).\
mode("overwrite").\
saveAsTable(f'{catalog}.silver.slv_calendar')
```
